## Handling Outliers in `actual_worth`

Initial inspection of `actual_worth` revealed a heavily right-skewed distribution:

- **Minimum**: 1 AED
- **Median**: ~1.36M AED
- **Mean**: ~3.88M AED (far above median, indicating skew)
- **Maximum**: ~13.8B AED

This skew is driven by two issues:

1. **Near-zero values** (387 rows under 10,000 AED) — likely gifts, family transfers, or data entry errors rather than genuine market sales.
2. **Extreme high-value outliers** — the 99th percentile sits at ~36M AED, but the 99.99th percentile reaches ~1.4B AED, suggesting a small number of large portfolio or land deals.

Left uncorrected, these outliers would distort YoY growth rates, area-level averages, and the Area Ranking model — a handful of mega-deals could make an entire area appear as a "growth star" in a single year.

### Approach

```python
# Drop clearly non-market transactions (near-zero value)
df = df[df['actual_worth'] >= 10000]

# Cap extreme outliers at the 99th percentile for analysis purposes
# (raw column is preserved; capped version used for growth/ranking calcs)
cap = df['actual_worth'].quantile(0.99)
df['actual_worth_capped'] = df['actual_worth'].clip(upper=cap)
```

- Rows below 10,000 AED are removed outright, as they don't represent real market activity.
- Values above the 99th percentile are **capped, not removed** — this preserves row counts (useful for transaction volume metrics) while preventing extreme values from distorting averages and growth calculations.
- The original `actual_worth` column is retained for metrics like **Total AED Value**, where the true (uncapped) sum matters.
- The new `actual_worth_capped` column should be used for **average price**, **YoY growth %**, and the **Area Ranking model**, where outlier sensitivity is a concern.

**Note:** Median (not mean) is still recommended for area-level price comparisons, since real estate prices are naturally skewed even after capping.

````python
# Drop clearly non-market transactions (near-zero value)
df = df[df['actual_worth'] >= 10000]

# Cap extreme outliers at 99th percentile for analysis purposes
# (keep the raw column, add a capped version for growth/ranking calcs)
cap = df['actual_worth'].quantile(0.99)
df['actual_worth_capped'] = df['actual_worth'].clip(upper=cap)
````

## 6. Outlier Treatment
`actual_worth` showed severe right-skew:
- Mean (3.88M AED) far exceeded median (1.36M AED)

**Approach:** rather than removing outliers outright, the original `actual_worth` was preserved for metrics requiring true totals (e.g. Total AED Value), while a new column `actual_worth_capped` was created by clipping values at the 99th percentile for use in growth rates, averages, and the Area Ranking model.

```python
cap = df['actual_worth'].quantile(0.99)

```

Skewness was confirmed visually via histogram and boxplot comparisons of raw vs. capped values.
